```python
def linear_combination(image1, image2, fraction1, fraction2):
  return image1 * fraction1 + image2 * fraction2
```

```python
plt.imshow(linear_combination(lena_gray, mandril_gray, 0.7, 0.3))
```

```python
plt.imshow(cv2.absdiff(lena_gray, mandril_gray))
```

```python
def hist(img):
  h = np.zeros((256, 1), np.float32) # creates and zeros single - column arrays
  height, width = img.shape[:2] # shape - we take the first 2 values
  for y in range(height):
    for x in range(width):
      h[img[y, x]] += 1
  return h
```

```python
plt.plot(hist(mandril_gray))
```

```python
plt.plot(cv2.calcHist ([mandril_gray], [0], None, [256], [0, 256]))
```

```python
eq_hist = cv2.equalizeHist(mandril_gray)
plt.plot(cv2.calcHist ([eq_hist], [0], None, [256], [0, 256]))
```

```python
clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
I_CLAHE = clahe.apply(mandril_gray)
plt.plot(cv2.calcHist ([I_CLAHE], [0], None, [256], [0, 256]))
# cv2.calcHist(images, channels, mask, histSize, ranges)
```

```python
gaussian_blur = cv2.GaussianBlur(mandril_gray, (15, 15), 0)
sobel = cv2.Sobel(mandril_gray, cv2.CV_64F, 1, 0, ksize=3)
lap = cv2.Laplacian(mandril_gray, cv2.CV_64F)
median = cv2.medianBlur(mandril_gray, 15)

figure, axis = plt.subplots(2, 2, figsize=(8, 8))

axis[0, 0].imshow(gaussian_blur)
axis[0, 1].imshow(sobel)
axis[1, 0].imshow(lap)
axis[1, 1].imshow(median)
```

```python
bilateral = cv2.bilateralFilter(mandril, 9, 75, 75)
kernel = cv2.getGaborKernel((21,21), 5, 0, 10, 0.5)
gabor = cv2.filter2D(mandril, -1, kernel)
erosion = cv2.erode(mandril, (3, 3))
```

```python
figure, axis = plt.subplots(1, 3, figsize=(12, 4))

axis[0].imshow(bilateral)
axis[1].imshow(gabor)
axis[2].imshow(erosion)
```
